```python
import csv

def convert_dat_to_csv(dat_file_path, csv_file_path):
    try:
        with open(dat_file_path, 'r') as dat_file:
            lines = dat_file.readlines()  # Read all lines from the .dat file

        with open(csv_file_path, 'w', newline='') as csv_file:
            writer = csv.writer(csv_file)
            for line in lines:
                # Split the line using '::' as the delimiter
                writer.writerow(line.strip().split('::'))
        
        print(f"File converted successfully: {csv_file_path}")
    except FileNotFoundError:
        print("The specified file was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
convert_dat_to_csv(movie_dat, movie_file)
convert_dat_to_csv(rating_dat, rating_file)
convert_dat_to_csv(user_dat, user_file)
```

```python
import pandas as pd

# Read the file with column names
movie_df = pd.read_csv(movie_file, encoding='ISO-8859-1', names=['Movie ID', 'Title', 'Genre'], header=None)

# Reset index to start from 1
movie_df.index = range(1, len(movie_df) + 1)

movie_df
```

```python
# Extract everything within brackets
bracketed_items = movie_df['Title'].str.extractall(r'\((.*?)\)')

# Count the total number of bracketed items
total_count = len(bracketed_items)

# Display the total count
print(f"Total number of bracketed items: {total_count}")
```

```python
# Extract the year from the title using a regular expression
movie_df['Year'] = movie_df['Title'].str.extract(r'\((\d{4})\)', expand=False)

# Remove the year from the Title column
movie_df['Title'] = movie_df['Title'].str.replace(r'\s*\(\d{4}\)', '', regex=True)

movie_df
```
